# Consulta ao DW

Exemplo de consulta analítica no modelo estrela (`public`) usando `obter_engine()`, `pandas.read_sql()` e `matplotlib`.

```python
from pathlib import Path
import sys

RAIZ = Path.cwd()
while not (RAIZ / ".git").exists() and RAIZ != RAIZ.parent:
    RAIZ = RAIZ.parent
if str(RAIZ) not in sys.path:
    sys.path.insert(0, str(RAIZ))

import pandas as pd
from sqlalchemy import text
from src.conexao import obter_engine

engine = obter_engine()
```

```python
# query de exemplo
query = """
SELECT
    t.ano,
    tr.modo_transporte_vitima,
    SUM(f.qtd_pessoas) AS vitimas,
    SUM(f.flag_obito)  AS obitos,
    ROUND(100.0 * SUM(f.flag_obito) / SUM(f.qtd_pessoas), 2) AS letalidade_pct
FROM public.fato_pessoa_sinistro AS f
JOIN public.dim_tempo              AS t  ON t.sk_tempo = f.sk_data_sinistro
JOIN public.dim_transporte         AS tr ON tr.sk_transporte = f.sk_transporte
GROUP BY t.ano, tr.modo_transporte_vitima
ORDER BY t.ano, obitos DESC;
"""

df = pd.read_sql(text(query), engine)
df
```

## 3. Leitura e Validação com Pandas

Testa a execução de query diretamente para DataFrame do Pandas usando a Engine configurada:

```python
import pandas as pd
from sqlalchemy import text

engine = obter_engine()

# Consulta as tabelas do schema public
query = """
SELECT table_schema, table_name
FROM information_schema.tables
WHERE table_schema = 'public'
ORDER BY table_name;
"""

df_tabelas = pd.read_sql(text(query), con=engine.connect())
print(f"Tabelas existentes no schema public: {len(df_tabelas)}")
df_tabelas
```

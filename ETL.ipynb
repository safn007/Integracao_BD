## Extraindo os dados

```python
import pandas as pd

arquivos = {}
for ano in [2021, 2022, 2023]:
    arquivos[ano] = [f"data/pessoas_{mes:02d}-{ano}.csv" for mes in range(1, 13)]

# dicionário: cada ano guarda a lista dos seus 12 arquivos
```

### Lendo e juntando os arquivos de cada ano

```python
dfs_por_ano = {}
falhas = []

for ano, caminhos in arquivos.items():
    lista = []
    for caminho in caminhos:
        try:
            df_mes = pd.read_csv(caminho, encoding='latin-1', sep=';')
        except (FileNotFoundError, pd.errors.ParserError, pd.errors.EmptyDataError) as e:
            falhas.append(caminho)
            print(f"Erro ao ler {caminho}: {e}")
            continue
        df_mes['ano_origem'] = ano
        df_mes['arquivo_origem'] = caminho
        lista.append(df_mes)

    if lista:                                   # evita concat de lista vazia
        dfs_por_ano[ano] = pd.concat(lista, ignore_index=True)
        print(ano, "->", len(lista), "arquivos,", len(dfs_por_ano[ano]), "linhas")

print("Arquivos com falha:", falhas)            # esperado: []

df = pd.concat(dfs_por_ano.values(), ignore_index=True)
print("Total de linhas no df unificado:", len(df))
```

## Transformação: Limpando os dados unificados

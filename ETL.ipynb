## Extraindo os dados

```python
import pandas as pd

arquivos = {}
for ano in [2021, 2022, 2023]:
    arquivos[ano] = [f"data/pessoas_{mes:02d}-{ano}.csv" for mes in range(1, 13)]

# dicionário: cada ano guarda a lista dos seus 12 arquivos
```

### Lendo e juntando os arquivos de cada ano

```python
dfs_por_ano = {}

for ano, caminhos in arquivos.items():
    lista = []
    for caminho in caminhos:
        try:
            df_mes = pd.read_csv(caminho, encoding='latin-1', sep=';') # usando o "sep = ';'" porque os arquivos CSV estão separados por ponto e vírgula
            df_mes['ano_origem'] = ano # adicionando a coluna 'ano_origem' com o valor do ano
            df_mes['arquivo_origem'] = caminho # adicionando a coluna 'arquivo_origem' com o valor do caminho
            lista.append(df_mes)
        except Exception as e: # se algum arquivo der erro, mostra qual foi e segue para o próximo
            print(f"Erro ao ler {caminho}: {e}")
    dfs_por_ano[ano] = pd.concat(lista, ignore_index=True)
    print(ano, "->", len(dfs_por_ano[ano]), "linhas")
```
